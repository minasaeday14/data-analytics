## Exercise 1.B NumPy Random

```python
import numpy as np
```

## Generating random numbers

```python
r_1 = np.random.randint(1, 21, size=2)
print(r_1)
```

```python
r_2 = np.random.randint(1,21, size=(1,2))
print(r_2)
```

```python
r_3 = np.random.random((2,1))
print(r_3)
```

```python
r_4 = np.random.randint(20,101, size=(4,5))
print(r_4)
```

```python
r_5 = np.random.random((4, 5))
print(r_5)
```

## Generating random choices from an array

```python
students = np.array(["jessica","cagla","rodas", "kidus",
    "sultan", "longley", "kiki", "angel",
    "tracy", "fuat", "onur", "shay"
])

print(students)
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3,4))
print(breakout_groups)

# This may not be the best way to assign breakout groups because
# the same student can appear more than once and some students may not appear at all.
```

```python
call_on_weighted = np.random.choice(
    students,
    size=5,
    p=[0.30, 0.10, 0.08, 0.07, 0.06, 0.05, 0.09, 0.00, 0.07, 0.06, 0.06, 0.06]
)
print(call_on_weighted)  
# p values:
# jessica = 0.30
# cagla = 0.10
# rodas = 0.08
# kidus = 0.07
# sultan = 0.06
# longley = 0.05
# kiki = 0.09
# angel = 0.00
# tracy = 0.07
# fuat = 0.06
# onur = 0.06
# shay = 0.06
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4,5))
print(bell_curve)
```

```python
print(np.mean(bell_curve))
```
