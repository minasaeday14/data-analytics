```python
import numpy as np
```

## Sample arrays

```python
r_1 = np.random.randint(1, 21, size=2)
r_2 = np.random.randint(1, 21, size=(1, 2))
r_3 = np.random.random((2, 1))
r_4 = np.random.randint(20, 101, size=(4, 5))
r_5 = np.random.random((4, 5))

print("r_1 =", r_1)
print("r_2 =\n", r_2)
print("r_3 =\n", r_3)
print("r_4 =\n", r_4)
print("r_5 =\n", r_5)
```

## Permutaions

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

The rows of `r_4` were shuffled in place. The array changed directly, but the values inside each row stayed together.


```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print(r_4)
```

I would use `shuffle` when I want to rearrange the original array itself. I would use `permutation` when I want a shuffled version without changing the original array.


## Mathematical Operations

```python
print("r_1 =", r_1)
print("shape of r_1 =", r_1.shape)

print("r_2 =\n", r_2)
print("shape of r_2 =", r_2.shape)

result_1_2 = np.add(r_1, r_2)
print("r_1 + r_2 =", result_1_2)
print("shape of result =", result_1_2.shape)
```

```python
print("r_1 =", r_1)
print("shape of r_1 =", r_1.shape)

print("r_3 =\n", r_3)
print("shape of r_3 =", r_3.shape)

result_1_3 = np.add(r_1, r_3)
print("r_1 + r_3 =\n", result_1_3)
print("shape of result =", result_1_3.shape)
```

```python
print("r_2 =\n", r_2)
print("shape of r_2 =", r_2.shape)

print("r_4 =\n", r_4)
print("shape of r_4 =", r_4.shape)

result_2_4 = np.add(r_2, r_4)
print("r_2 + r_4 =\n", result_2_4)
print("shape of result =", result_2_4.shape)
```

```python
# r_1 + r_4 causes an error because their shapes are not compatible for broadcasting
np.add(r_1, r_4)
```

```python
# r_3 + r_4 also causes an error because their shapes are not compatible for broadcasting
np.add(r_3, r_4)
```

```python
# r_2 + r_4 causes an error because (1, 2) and (4, 5) do not match for broadcasting
np.add(r_2, r_4)
```

```python
print("r_4 =\n", r_4)
print(np.add(np.array([1]), r_4))
```

```python
print(np.add(np.array([[1]]), r_4))
```

```python
print(np.add(np.array([1, 1, 1, 1, 1]), r_4))
```

```python
# This produces an error because (2, 1) cannot broadcast to (4, 5)
np.add(np.array([[1], [1]]), r_4)
```

Broadcasting is NumPy's way of making arrays with different shapes work together when possible. It compares shapes from right to left. Two dimensions are compatible if they are equal or if one of them is 1. When using `add`, NumPy adds matching values position by position after stretching any dimension of size 1 as needed. The resulting array shape is based on the broadcasted shape that both arrays can fit into.
