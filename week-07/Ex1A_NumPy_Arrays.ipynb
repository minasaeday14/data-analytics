```python
import numpy as np
```

## Looping through arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop goes through the first dimension of the array, so it works with one full row at a time and prints each row multiplied by 2. The second loop goes through the second dimension, so it works with one individual number at a time and prints each value multiplied by 2 separately.


```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# This result is different because x represents a full row, not a single number.
# So x * 2 doubles the whole row each time, and since it is inside the inner loop,
# the same doubled row gets printed once for each value in that row.
```

## Joining and Stacking arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1,joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)
# This gives an error because the two arrays do not have the same shape.
# np.stack() requires all input arrays to be exactly the same size and dimensions.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
print(another_2)

# This gives an error because the arrays do not have the same number of rows.
# To concatenate on axis=1, the arrays must match in size along axis=0
```

```python
horizontal_stack = np.hstack((arrA,arrA))
print(horizontal_stack) 
# hstack joins the arrays side by side by adding more columns.

vertical_stack = np.vstack((arrA,arrA))
print(vertical_stack)     

# vstack joins the arrays on top of each other by adding more rows.

depth_stack = np.dstack((arrA, arrA))
print(depth_stack)  

# dstack stacks the arrays along a third dimension, adding depth.
```

## Splititng arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)

# This creates 2 arrays.
# Each result is still a 2-dimensional array.
# Since joined_array_1 has 4 rows, each split gets 2 rows.
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# This creates 3 arrays.
# Each result is still a 2-dimensional array.
# Because 4 rows cannot be split evenly into 3 parts
```

## Searching and Replacing Arrays Values

```python
result_1 = np.where(arrA % 2 == 0, arrA,  'odd')
print(result_1)
```
